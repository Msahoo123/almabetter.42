## **Activities performed**
## **1. Load the dataset**

The dataset is loaded into Python using the Pandas library. The data may be stored in CSV format.

##**4.4 Stage 2: Data Preprocessing**
Data preprocessing is an important stage because raw datasets may contain missing values, duplicate records, inconsistent categories, or unsuitable data types.
The objective is to prepare clean and consistent data for machine learning.

**Activities performed**

**1. Handle missing values**

Missing values are identified and handled using appropriate techniques.
For numerical columns, median imputation may be suitable. For categorical columns, the most frequent category can be used.

```python
# Identify missing values
print(df.isnull().sum())

# Fill numerical missing values with the median
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

Categorical missing values should also be handled before model training.

**2. Remove duplicate records**

Duplicate rows are checked and removed when appropriate.

```python
df = df.drop_duplicates()

print("Dataset shape:", df.shape)
```

**3. Encode categorical data**

Machine learning models generally require numerical inputs. Categorical features such as gender, travel type, and class can be converted into numerical representations using one-hot encoding.

```python
df_encoded = pd.get_dummies(
    df,
    columns=["Gender", "Type of Travel", "Class"],
    drop_first=True
)
```

The column names must match those in the actual dataset.

**4. Separate features and target**

The input features are separated from the target variable.

```python
X = df_encoded.drop(columns=["Target"])
y = df_encoded["Target"]
```

Here, X contains passenger and travel information, while y contains the recommendation outcome.

**5. Split the dataset**

The data is divided into training and testing sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

The training set is used to fit the model, while the testing set is used to evaluate its performance on unseen records.

**Important:** In a production-quality workflow, missing-value imputation, encoding, and scaling should be fitted using training data only. A Scikit-learn pipeline helps prevent data leakage.

**Expected outcome:** Clean, appropriately encoded data that is ready for analysis and model training.

## **4.6 Stage 4: Model Building and Evaluation**

The fourth stage involves training machine learning classification algorithms to predict whether a passenger will recommend IndiGo Airlines.
Different models are compared to identify a suitable model for the dataset.

**1. Logistic Regression**

Logistic Regression is a classification algorithm that estimates the probability of a passenger belonging to a particular class.
It provides a useful baseline model for binary classification.

**2. Decision Tree**

A Decision Tree learns decision rules from the input features. It can identify patterns involving passenger ratings, travel details, and delays.
It is relatively easy to interpret, although a deep tree can overfit the training data.

**3. Random Forest**

Random Forest combines multiple decision trees to produce predictions.
It can capture complex relationships between passenger features and recommendations. It also provides feature-importance scores that can help identify variables useful for prediction.

**4. Gradient Boosting**

Gradient Boosting builds a sequence of models, with each new model attempting to improve upon errors made by previous models.
It can perform well on structured datasets when appropriately tuned.

**Example Python code**

The following is a basic example of training and evaluating a Random Forest classifier. It assumes that X_train, X_test, y_train, and y_test have already been prepared and that categorical variables have been encoded.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Create the model
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate performance
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_pred, zero_division=0))
print("F1-score:", f1_score(y_test, y_pred, zero_division=0))

print(classification_report(y_test, y_pred, zero_division=0))
```

**Model evaluation metrics**

| Metric | Purpose |
|---|---|
| Accuracy | Measures the proportion of correct predictions. |
| Precision | Measures how many predicted referrals are actually positive. |
| Recall | Measures how many actual positive referrals the model identifies. |
| F1-score | Balances precision and recall. |
| ROC-AUC | Measures how effectively the model ranks positive cases above negative cases across classification thresholds. |

Hyperparameter tuning can also be performed using cross-validation to identify suitable model settings.

**Expected outcome:** A trained classification model and evaluation results that show how well it predicts passenger recommendations.

##**5.2 Referral Distribution**

**Objective**

The purpose of this analysis is to understand how many passengers recommend IndiGo Airlines and how many do not.

The referral distribution is the target-variable analysis of the project.
- No: Passengers who do not recommend the airline.
- Yes: Passengers who recommend the airline.

According to the example visualization, approximately 68% of passengers belong to the No category, while 32% belong to the Yes category.

**Python Code**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("indigo_reviews.csv")

# Display target distribution
print(df["Target"].value_counts())

# Calculate percentages
print(df["Target"].value_counts(normalize=True) * 100)

# Create visualization
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Target")

plt.title("Passenger Referral Distribution")
plt.xlabel("Passenger Referral")
plt.ylabel("Number of Passengers")

plt.show()
```

Replace Target with the actual name of your recommendation column.

## **Analysis and Interpretation**

The example graph indicates that more passengers do not recommend the airline than recommend it.

This difference suggests that the dataset may have an imbalanced target variable. Class imbalance is important because a machine learning model may achieve high accuracy simply by predicting the majority class.

For example, if 68% of passengers belong to the No category, a model that always predicts No could achieve approximately 68% accuracy without correctly identifying any positive referrals.

Therefore, I will evaluate the model using precision, recall, F1-score, and ROC-AUC alongside accuracy.

**Key Findings**

- The No category represents approximately 68% of passengers in the example visualization.
- The Yes category represents approximately 32%.
- The target distribution should be considered when selecting and evaluating machine learning models.
- The airline can investigate the reasons behind negative recommendations.

**Business insight:** Understanding the distribution of recommendations helps identify the scale of passenger dissatisfaction and supports further investigation into service improvements.

##**5.3 Referral by Class**

**Objective**

This analysis examines whether passenger recommendations differ across travel classes.

The dataset includes three travel classes:
1. Economy
2. Economy Plus
3. Business

Passengers may have different expectations depending on their travel class. For example, they may evaluate seat comfort, service quality, convenience, and value for money differently.

**Python Code**

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Class",
    hue="Target"
)

plt.title("Passenger Referral by Travel Class")
plt.xlabel("Travel Class")
plt.ylabel("Passenger Count")

plt.legend(title="Referral")
plt.tight_layout()
plt.show()
```

## **Analysis and Interpretation**

The graph compares positive and negative recommendations within each travel class.

The example visualization shows that Economy has the largest number of passengers, followed by Economy Plus and Business.

However, the number of passengers in a class does not necessarily indicate its referral rate. Economy may have more positive referrals simply because it contains more passengers.

To compare the classes fairly, I will calculate the percentage of passengers who recommend the airline within each class.

## **Python Code: Referral Rate by Class**

```python
referral_rate = (
    df.groupby("Class")["Target"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

print(referral_rate)

referral_rate.plot(kind="bar", figsize=(8, 5))

plt.title("Referral Rate by Travel Class")
plt.xlabel("Travel Class")
plt.ylabel("Referral Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

- Are business-class passengers more likely to recommend the airline?
- Which service improvements might be appropriate for each class?

##**6.3 Model 1: Logistic Regression**

Logistic Regression is a supervised machine learning algorithm used for classification problems. Despite the word regression in its name, it is commonly used to predict categories such as Yes/No, Pass/Fail, and Positive/Negative.

In this project, Logistic Regression predicts whether a passenger is likely to recommend IndiGo Airlines.

Logistic Regression learns the relationship between these input features and the recommendation outcome.

###**How Logistic Regression Works**

The algorithm calculates a score from the input features and converts that score into a probability between 0 and 1.

The logistic function is:

                     P(Y=1∣X)=1/(1+e^(-z) )




Where:
- \(P(Y=1\mid X)\) is the predicted probability that the passenger will recommend the airline.
- \(X\) represents the passenger information and service ratings.
- \(z\) is a weighted combination of the input features.
- \(e\) is the mathematical constant used in the logistic function.

For example, suppose the model predicts a recommendation probability of 0.80. If the classification threshold is 0.50, the model classifies the passenger as likely to recommend the airline.

The actual probability and classification depend on the trained model and the passenger's data.

**Advantages**
- Easy to implement and train.
- Provides a useful baseline for classification.
- Can produce probabilities for individual passengers.
- Relatively easy to interpret.
- Supports regularization to help control overfitting.

**Limitations**

- May not capture complex nonlinear relationships without suitable feature transformations.
- Can be affected by multicollinearity among input features.

- Its performance depends on the quality of the input features and preprocessing.

##**6.4 Model 2: Decision Tree Classifier**

A Decision Tree is a supervised machine learning algorithm that makes predictions by following a sequence of conditions. It resembles a flowchart in which each internal node represents a decision and each leaf node represents a predicted class.

In the IndiGo passenger referral project, the model learns rules based on passenger information and service ratings.

For example, a simplified decision path might look like this:
- Is the passenger's overall service experience positive?
  - Yes: Check seat comfort and online boarding ratings.
  - No: Check whether delays or poor service ratings are associated with a negative recommendation.
- Based on the learned conditions, predict Yes or No.

These are illustrative examples, not actual rules learned from the dataset.

**How a Decision Tree Works**

The algorithm examines possible splits in the training data and selects conditions that help separate the target classes.

Common criteria include:

**Gini impurity:** Measures how mixed the classes are within a node.

\[
Gini=1-\sum_{i=1}^{k}p_i^2
\]

Entropy: Measures the uncertainty in the class distribution.

\[
Entropy=-\sum_{i=1}^{k}p_i\log_2(p_i)
\]

The algorithm attempts to create useful splits that improve class separation. It continues until a stopping condition is reached, such as a maximum tree depth or a minimum number of samples in a leaf.

**Advantages**

- Easy to understand and visualize.
- Can model nonlinear relationships.
- Can capture interactions between features.
- Does not require numerical feature scaling.
- Can work with both numerical and categorical features after appropriate encoding.

**Limitations**
- A deep tree can overfit the training data.
- Small changes in the training dataset may produce a different tree.
- A single tree may generalize less effectively than an ensemble.
- Unrestricted growth can produce a complicated model.

**Result in This Project**

The Decision Tree achieved 84% reported accuracy, improving on Logistic Regression by two percentage points.

Its main advantage is interpretability: the learned rules can help analysts investigate which passenger characteristics are associated with recommendations.

##**6.5 Model 3: Random Forest Classifier**

Random Forest is an ensemble learning algorithm that combines multiple decision trees to produce a final prediction.

Instead of relying on one tree, it trains many trees using randomized samples of the training data and randomized subsets of features. For a classification problem, the trees' predictions are combined, typically through majority voting or averaged class probabilities.

In this project, Random Forest uses passenger demographics, travel details, service ratings, and delay information to predict whether passengers are likely to recommend IndiGo Airlines.

##**How Random Forest Works**

The model follows these steps:

**Step 1: Prepare the dataset**

Clean missing values, encode categorical features, separate the target variable, and split the data into training and testing sets.

**Step 2: Create training samples**

Random Forest trains each tree on a bootstrap sample of the training data.

**Step 3: Select features**

At each split, a tree considers a randomized subset of features rather than every feature.

**Step 4: Train multiple decision trees**

Each tree learns patterns from its training sample and selected features.

**Step 5: Combine predictions**

For classification, the model combines the trees' predictions to determine the final class or recommendation probability.

**Example**

Suppose a Random Forest contains 100 decision trees. For one passenger:
- 75 trees predict Yes.
- 25 trees predict No.

The majority vote predicts Yes.

This example illustrates how voting works; it does not describe the actual number of trees or predictions from the reported project results.

##**Why Random Forest Is Suitable for This Project**

Passenger referral decisions can depend on many factors at the same time. For example, a passenger's recommendation may be associated with a combination of travel class, seat comfort, online boarding, flight delays, and in-flight services.

Random Forest can learn nonlinear patterns and interactions between features without requiring the analyst to specify every possible relationship manually.

It can also provide feature-importance estimates that help identify variables useful for prediction. These estimates should be interpreted carefully because correlated features can share or distort apparent importance.

##**Advantages**
- Can learn complex patterns.
- Combines multiple trees instead of relying on one tree.
- Often generalizes better than an unrestricted single decision tree.
- Can provide class probabilities and feature-importance estimates.
- Usually does not require feature scaling.

##**Limitations**

##**6.6 Model 4: Gradient Boosting Classifier**

Gradient Boosting is an ensemble machine learning technique that builds a sequence of relatively small predictive models, commonly decision trees.

Each new model attempts to improve the combined model by focusing on errors or remaining predictive patterns from earlier stages.

In the IndiGo project, Gradient Boosting can learn relationships between passenger recommendations and factors such as service ratings, travel class, flight distance, and departure or arrival delays.

##**How Gradient Boosting Works**

The algorithm operates through a sequence of steps:
1. Start with an initial prediction.
2. Measure how far the predictions are from the actual target values using a chosen loss function.
3. Train a new decision tree to improve the current predictions.
4. Add the new tree's contribution to the existing ensemble.
5. Repeat the process for a specified number of boosting stages.
6. Use the combined model to predict the recommendation outcome.
The learning rate controls how much each new tree contributes to the ensemble.

##**Example**
Suppose the initial model incorrectly predicts the recommendation outcome for some passengers. A subsequent tree attempts to reduce the remaining prediction error. Additional trees continue improving the model until the chosen number of boosting stages is reached.

This process can help the model capture complex relationships in the passenger dataset.

##**Advantages**
- Can model nonlinear relationships.
- Can capture interactions between passenger features.
- Often performs well on structured data.
- Can provide useful predictive performance after appropriate tuning.

##**Limitations**
- Training may take longer than training a single tree.
- Performance depends on hyperparameters such as learning rate and tree depth.
- Too many trees or overly complex trees can lead to overfitting.
- The resulting model is more difficult to interpret than a simple decision tree.

##**Result in This Project**

This result indicates strong reported classification performance. However, whether it is the better choice depends on the model's other evaluation metrics, computational requirements, and reliability on new data.

##**6.7 Model 5: XGBoost Classifier**

For this project, XGBoost can learn patterns in passenger demographics, service ratings, travel information, and delay variables to predict whether a passenger is likely to recommend IndiGo Airlines.

##**How XGBoost Works**

XGBoost builds an ensemble of decision trees sequentially. Each new tree improves the existing model by learning patterns that the current model has not captured adequately.

It also includes regularization mechanisms that penalize unnecessary model complexity. These mechanisms can help reduce overfitting when configured appropriately.

The general workflow is:
1. Prepare and preprocess the dataset.
2. Train an initial boosting model.
3. Calculate the remaining prediction errors.
4. Add trees that improve the model's predictions.
5. Use regularization and hyperparameter settings to control complexity.
6. Evaluate predictions on unseen test data.

##**Important Hyperparameters**

| Hyperparameter | Meaning | Purpose |

These settings should be tuned using cross-validation or another suitable validation strategy rather than selected solely from the test-set accuracy.

##**Advantages**

##**Limitations**

##**6.8 Python Code: Training and Comparing All Five Models**

The following Python code trains the five models, evaluates their accuracy on a held-out test set, and displays a comparison chart.

**Important:** This code calculates new results from the CSV file. It does not force the accuracy values to match the example table. Your actual results may differ.

The example assumes your CSV file is named indigo_reviews.csv and that the target column is named Target, containing Yes and No. Change these names if your dataset uses different column names.

##**Step 1: Import Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
```

If XGBoost is not installed in your environment, install it in a notebook cell:

```python
%pip install xgboost
```

##**Step 2: Load the Dataset**

```python
df = pd.read_csv("indigo_reviews.csv")

print("Dataset Shape:", df.shape)
print(df.head())
print(df.info())
```

This step loads the passenger dataset and displays its structure.

Check the column names and target values:

```python
print("Column Names:")
print(df.columns.tolist())

print("\nTarget Distribution:")
print(df["Target"].value_counts(dropna=False))
```

Make sure the target column really is named Target. The code must be adjusted if the CSV uses a different name.

##**Step 3: Prepare the Features and Target**

```python
# Remove rows where the target is missing
df = df.dropna(subset=["Target"]).copy()

# Input features
X = df.drop(columns=["Target"])

# Target variable
y = df["Target"].astype(str).str.strip()

# Keep only rows with recognized target labels
valid = y.isin(["Yes", "No"])
X = X.loc[valid].copy()
y = y.loc[valid].map({"No": 0, "Yes": 1})

# Identify numerical and categorical columns
numeric_features = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("Numerical Features:", numeric_features)
print("Categorical Features:", categorical_features)
```

Here, No is encoded as 0 and Yes as 1. This allows the models to predict the two referral classes.

If the dataset uses values such as Recommended and Not Recommended, change the mapping to match the actual target labels.

##**Step 4: Split the Dataset**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

The dataset is divided into:
- 80% training data: Used to learn patterns.
- 20% testing data: Used to evaluate the trained models.

The stratify=y argument attempts to preserve the proportion of Yes and No labels in both sets.

If multiple rows belong to the same passenger, or if the dataset has a time-based structure, a random split may not be appropriate. A passenger-level or time-based split may be necessary to avoid information leakage.

##**Step 5: Create the Preprocessing Pipeline**

```python
numeric_preprocessor = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_preprocessor = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_preprocessor, numeric_features),
        ("cat", categorical_preprocessor, categorical_features)
    ]
)
```

This preprocessing pipeline:
- Fills missing numerical values with the training-set median.
- Scales numerical variables.
- Fills missing categorical values with the most frequent category.
- Converts categorical variables into numerical indicator columns.

Because the preprocessing pipeline is fitted within each model pipeline, the imputation and scaling parameters are learned from the training data rather than the test data.

##**Step 6: Define the Models**

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=10,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=None,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
}
```

These are example settings, not guaranteed optimal hyperparameters. Actual model performance depends on the dataset and the chosen settings.

##**Step 7: Train and Evaluate the Models**

```python
results = []
trained_models = {}

for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("classifier", model)
        ]
    )

    # Train the model
    pipeline.fit(X_train, y_train)

    # Predict test labels
    y_pred = pipeline.predict(X_test)

    # Predict probabilities for the positive class
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )
    auc = roc_auc_score(y_test, y_prob)

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": auc
    })

    trained_models[name] = pipeline

    print(f"\n{name}")
    print(classification_report(
        y_test,
        y_pred,
        target_names=["No", "Yes"],
        zero_division=0
    ))

results_df = pd.DataFrame(results)

print("\nModel Performance Comparison:")
print(
    results_df
    .sort_values("Accuracy", ascending=False)
    .round(3)
    .to_string(index=False)
)
```

This code reports accuracy, precision, recall, F1-score, and ROC-AUC for each model.

These additional metrics help determine whether a model performs well on both recommendation classes rather than merely achieving a high overall accuracy.

##**Step 8: Plot the Model Accuracy Comparison**

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=results_df.sort_values("Accuracy", ascending=False),
    x="Accuracy",
    y="Model"
)

plt.title("Machine Learning Model Accuracy Comparison")
plt.xlabel("Accuracy")
plt.ylabel("Machine Learning Model")
plt.xlim(0, 1)

plt.tight_layout()
plt.show()
```

##**6.9 Understanding the Evaluation Metrics**

Although accuracy is useful, it should not be the only measure used to select the final model.

**1. Accuracy**

Accuracy measures the proportion of all test predictions that are correct.



\[
Accuracy=\frac{TP+TN}{TP+TN+FP+FN}
\]

For example, 87% accuracy means the model correctly classified approximately 87% of the test observations, assuming that result was calculated on the stated test set.

##**2. Precision**

Precision measures how many passengers predicted as Yes actually belong to the Yes class.

\[
Precision=\frac{TP}{TP+FP}
\]

High precision means that fewer passengers are incorrectly classified as likely to recommend the airline.

##**3. Recall**

Recall measures how many actual Yes passengers are correctly identified.

\[
Recall=\frac{TP}{TP+FN}
\]

High recall means that the model identifies a larger proportion of passengers who actually recommend the airline.

The appropriate priority depends on the business objective. For identifying dissatisfied passengers, recall for the No class may be more useful than recall for Yes.

##**4. F1-Score**

The F1-score combines precision and recall into a single measure using their harmonic mean.

\[
F1=2\times\frac{Precision\times Recall}{Precision+Recall}
\]

It can be particularly useful when the target classes are imbalanced and both precision and recall matter.

##**5. ROC-AUC**

ROC-AUC measures how well the model ranks positive examples above negative examples across different classification thresholds.

A higher ROC-AUC generally indicates better separation between the two classes, although the practical usefulness of the model still depends on the business objective and the selected threshold.

##**7.6 F1-Score — 84%**

**What Is the F1-Score?**

The F1-score combines precision and recall into a single metric.

Precision measures the reliability of positive predictions, while recall measures the proportion of actual positive cases identified.

The F1-score provides a balance between these two measures by calculating their harmonic mean.

It is particularly useful when the target classes are imbalanced or when both false positives and false negatives matter.

**F1-Score Formula**

                     F1=2×(Precision×Recall)/(Precision+Recall)


Where:
- Precision measures the reliability of positive predictions.
- Recall measures the proportion of actual positive cases correctly identified.

**Calculation Using the Reported Values**

The reported precision is 0.85, and the reported recall is 0.83.

Substituting these values into the formula:

                      F1=2×(0.85×0.83)/(0.85+0.83)

                      F1 = 1.411/1.68
                      
                      F1≈0.840

This agrees with the reported F1-score of 84%, allowing for rounding.

###**Interpretation for This Project**

It suggests that the model achieves a useful combination of identifying actual Yes passengers and avoiding incorrect Yes predictions.

For the airline, this is important because relying only on precision could encourage the model to make fewer positive predictions, while relying only on recall could encourage it to classify more passengers as positive.

The F1-score helps evaluate both aspects together.

However, the F1-score does not include true negatives directly. If performance on both classes is important, it is useful to examine class-specific metrics and the confusion matrix as well.

**Business Importance**

The F1-score can help the airline:
- Compare classification models when both precision and recall matter.

##**7.7 ROC-AUC — 90%**

**What Is ROC-AUC?**

ROC-AUC stands for Receiver Operating Characteristic — Area Under the Curve.

It measures how well a model distinguishes between two classes across different classification thresholds.

Unlike accuracy, which measures predictions at a particular threshold, ROC-AUC evaluates how well the model ranks positive cases above negative cases across a range of thresholds.

For this project, the two classes are:
- Yes: The passenger recommends IndiGo Airlines.
- No: The passenger does not recommend IndiGo Airlines.

###**Understanding the ROC Curve**

The ROC curve plots two measures:

**True Positive Rate (TPR):**


                      TPR=TP/(TP+FN)

This is equivalent to recall for the positive class.

**False Positive Rate (FPR):**

                      FPR=FP/(FP+TN)

This measures the proportion of actual negative cases incorrectly classified as positive at a given threshold.

The ROC curve displays the true positive rate against the false positive rate at different thresholds.

The area under this curve is called ROC-AUC.

**Interpretation of ROC-AUC Values**

The following table provides a general guide rather than strict universal cutoffs.

| ROC-AUC Value | General Interpretation |
|---:|---|
| 0.50 | No ranking ability beyond chance |
| 0.60–0.70 | Limited discrimination |
| 0.70–0.80 | Moderate discrimination |
| 0.80–0.90 | Good discrimination |
| Above 0.90 | Very strong discrimination |

These categories are approximate. Whether a score is good enough depends on the dataset, the application, and the consequences of incorrect predictions.

###**Interpretation for This Project**

The reported ROC-AUC is 0.90.

This indicates strong discrimination between the Yes and No classes, assuming the value was calculated correctly using predicted probabilities on held-out test data.

A ROC-AUC of 0.90 can be interpreted as approximately a 90% chance that a randomly selected positive observation receives a higher model score than a randomly selected negative observation, subject to the usual treatment of tied scores.

This does not mean that the model makes correct predictions 90% of the time. That is the role of accuracy.

ROC-AUC also does not guarantee that the model's predicted probabilities are well calibrated. Calibration should be evaluated separately if the airline needs reliable probability estimates.

**Business Importance**

ROC-AUC can help the airline:
- Compare how effectively different models distinguish the two classes.
- Evaluate ranking performance independently of one fixed classification threshold.
- Choose an appropriate threshold for a particular business goal.
- Understand whether the model can separate passengers with different recommendation outcomes.

**Conclusion:** The reported ROC-AUC of 0.90 suggests strong class-discrimination ability, provided the result is confirmed on the actual test dataset.

##**7.9 Python Code to Calculate All Evaluation Metrics**

The following code calculates the five reported evaluation metrics from the predictions and predicted probabilities of a trained binary classification model.

It assumes that:

- best_model is the trained scikit-learn pipeline.
- X_test contains the test features.
- y_test contains labels encoded as 0 for No and 1 for Yes.

```python
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Predict class labels
y_pred = best_model.predict(X_test)

# Predict probabilities for the positive class
y_prob = best_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
metrics = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(
        y_test, y_pred, zero_division=0
    ),
    "Recall": recall_score(
        y_test, y_pred, zero_division=0
    ),
    "F1-Score": f1_score(
        y_test, y_pred, zero_division=0
    ),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

# Display the evaluation results
metrics_df = pd.DataFrame(
    list(metrics.items()),
    columns=["Evaluation Metric", "Value"]
)

metrics_df["Percentage"] = (
    metrics_df["Value"] * 100
).round(2)

print(metrics_df.to_string(index=False))
```

##**8.3 Random Forest Feature Importance**


Random Forest is an ensemble machine learning algorithm that combines predictions from multiple decision trees.

Each tree learns patterns from a randomized sample of the training data and a randomized subset of features. The trees' predictions are then combined to produce the final classification.

In this project, the Random Forest model predicts whether a passenger is likely to recommend IndiGo Airlines.

After training the model, we can examine feature importance to understand which input variables contributed most to the model's learned decision patterns.

**How Random Forest Calculates Feature Importance**

A common approach in scikit-learn's Random Forest implementation is mean decrease in impurity.

When a decision tree splits a node, it attempts to separate the target classes more effectively. A split that reduces impurity contributes to the importance of the feature used for that split.

The Random Forest combines the importance contributions across its trees and normalizes them so that the feature-importance scores sum to approximately 1.

In simplified terms:



     I_j=("Total weighted impurity reduction from feature " j)/"Total weighted impurity reduction from all features"

Where:
- \(I_j\) is the importance score of feature \(j\).
- The numerator represents the feature's total contribution to impurity reduction.
- The denominator represents the corresponding contribution from all features.

The precise calculation includes the weighting and normalization used by the implementation.

##**8.7 Alternative Method: Permutation Feature Importance**

Random Forest's built-in importance scores are convenient, but they have limitations. For example, impurity-based importance can favor certain types of features and may distribute importance unevenly among correlated variables.

An alternative is permutation importance.

This method measures how much a model's evaluation score changes when the values of one feature are randomly shuffled while the other features remain unchanged.

If shuffling a feature substantially reduces the model's performance, that feature may be important to the model's predictions on the evaluation dataset.

**Python Code**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.inspection import permutation_importance

# Calculate permutation importance on held-out test data
permutation_result = permutation_importance(
    best_model,
    X_test,
    y_test,
    scoring="accuracy",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

permutation_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": permutation_result.importances_mean,
    "Std": permutation_result.importances_std
})

permutation_df = permutation_df.sort_values(
    by="Importance",
    ascending=False
)

print(permutation_df.head(10))

# Plot the top five features
top_permutation = permutation_df.head(5).sort_values(
    by="Importance",
    ascending=True
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_permutation,
    x="Importance",
    y="Feature"
)

plt.title("Top 5 Features - Permutation Importance")
plt.xlabel("Mean Decrease in Accuracy After Permutation")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

**How to Interpret Permutation Importance**

If shuffling Inflight Wifi Service causes a substantial reduction in test accuracy, the model may rely on that feature for its predictions.

If shuffling a feature has almost no effect, the model may not gain much predictive value from that feature under the chosen evaluation conditions.

Permutation importance can also be misleading when features are strongly correlated. If two features provide similar information, the model may continue to perform well when either one is shuffled because the other still contains related information.

For this reason, it is useful to compare multiple importance methods and examine the underlying data.

##**8.8 Important Limitations of Feature Importance**

Feature importance should be interpreted carefully for the following reasons.

**1. Importance Does Not Prove Causation**

A feature with a high importance score is useful for prediction according to the selected method. It does not prove that changing that feature will cause the recommendation outcome to change.

For example, a high score for seat comfort would not by itself prove that improving seats will increase the number of passengers who recommend the airline.

**2. Correlated Features Can Share Importance**

Passenger service ratings may be correlated. For example, passengers who rate on-board service positively may also rate food and drink positively.

When features contain overlapping information, their importance scores may be shared or distributed in unexpected ways.

**3. Results Depend on the Dataset**

Feature importance depends on the passengers included in the dataset, the quality of the data, the target definition, preprocessing, and the trained model.

A different sample of passengers may produce a different ranking.

**4. Importance Is Not the Same as Satisfaction**

A feature can be important for prediction without having a high average rating.

Similarly, a feature with a high average rating may have low predictive importance if it does not help distinguish between the Yes and No classes.

**5. Validate Business Findings Separately**

If the airline wants to decide which service area to improve, feature importance should be combined with passenger feedback, operational data, cost analysis, and suitable experiments.

This helps prevent business decisions from relying on a predictive score alone.

###**Step 7: Analyze Departure Delays**

```python
delay_df = df.copy()

delay_df["Referral"] = (
    delay_df["Target"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({"yes": 1, "no": 0})
)

delay_df = delay_df.dropna(
    subset=["Referral", "Departure Delay in Minutes"]
)

delay_df["Departure Delay Group"] = pd.cut(
    delay_df["Departure Delay in Minutes"],
    bins=[-0.01, 0, 15, 60, float("inf")],
    labels=[
        "No Delay",
        "1–15 Minutes",
        "16–60 Minutes",
        "Over 60 Minutes"
    ]
)

departure_delay_analysis = (
    delay_df.groupby(
        "Departure Delay Group",
        observed=False
    )
    .agg(
        Referral_Rate=("Referral", "mean"),
        Passenger_Count=("Referral", "size")
    )
    .reset_index()
)

departure_delay_analysis["Referral Rate (%)"] = (
    departure_delay_analysis["Referral_Rate"] * 100
)

print(departure_delay_analysis)
```

This code groups departure delays into four categories and calculates the referral rate for each category.
